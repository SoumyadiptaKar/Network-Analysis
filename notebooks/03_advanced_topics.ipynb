# Advanced Graph Topics

This notebook covers advanced network analysis topics including algorithms, network models, and real-world applications.

```python
import networkx as nx
import matplotlib.pyplot as plt
import numpy as np
from collections import defaultdict
```

## Network Flow Algorithms

### Maximum Flow

Find the maximum flow from source to sink in a flow network.

```python
# Create a flow network
G = nx.DiGraph()

# Add edges with capacities
G.add_edge('S', 'A', capacity=10)
G.add_edge('S', 'B', capacity=5)
G.add_edge('A', 'C', capacity=15)
G.add_edge('A', 'B', capacity=4)
G.add_edge('B', 'C', capacity=8)
G.add_edge('B', 'D', capacity=4)
G.add_edge('C', 'T', capacity=10)
G.add_edge('D', 'T', capacity=10)

# Calculate maximum flow
flow_value, flow_dict = nx.maximum_flow(G, 'S', 'T')
print(f"Maximum flow value: {flow_value}")
print("\nFlow on each edge:")
for source in flow_dict:
    for target, flow in flow_dict[source].items():
        if flow > 0:
            print(f"  {source} -> {target}: {flow}")
```

```python
# Visualize the flow network
plt.figure(figsize=(10, 6))
pos = nx.spring_layout(G, seed=42)

nx.draw(G, pos, with_labels=True, node_color='lightblue',
        node_size=1000, font_size=12, font_weight='bold',
        edge_color='gray', width=2, arrows=True, arrowsize=20)

# Draw edge labels with capacities
edge_labels = nx.get_edge_attributes(G, 'capacity')
nx.draw_networkx_edge_labels(G, pos, edge_labels)

plt.title("Flow Network", fontsize=16, fontweight='bold')
plt.axis('off')
plt.tight_layout()
plt.show()
```

### Minimum Cut

Find the minimum cut that separates source from sink.

```python
# Find minimum cut
cut_value, partition = nx.minimum_cut(G, 'S', 'T')
reachable, non_reachable = partition

print(f"Minimum cut value: {cut_value}")
print(f"Source side: {sorted(reachable)}")
print(f"Sink side: {sorted(non_reachable)}")
```

## Shortest Path Algorithms

### Dijkstra's Algorithm

Find shortest paths in weighted graphs.

```python
# Create a weighted graph
G_weighted = nx.Graph()

edges = [
    ('A', 'B', 4), ('A', 'C', 2),
    ('B', 'C', 1), ('B', 'D', 5),
    ('C', 'D', 8), ('C', 'E', 10),
    ('D', 'E', 2), ('D', 'F', 6),
    ('E', 'F', 3)
]

G_weighted.add_weighted_edges_from(edges)

# Find shortest path
source = 'A'
target = 'F'

path = nx.shortest_path(G_weighted, source, target, weight='weight')
path_length = nx.shortest_path_length(G_weighted, source, target, weight='weight')

print(f"Shortest path from {source} to {target}:")
print(f"  Path: {' -> '.join(path)}")
print(f"  Total weight: {path_length}")
```

```python
# Visualize with shortest path highlighted
plt.figure(figsize=(10, 6))
pos = nx.spring_layout(G_weighted, seed=42)

# Draw all edges
nx.draw_networkx_edges(G_weighted, pos, edge_color='lightgray', width=2)

# Highlight shortest path
path_edges = list(zip(path, path[1:]))
nx.draw_networkx_edges(G_weighted, pos, edgelist=path_edges, 
                       edge_color='red', width=4)

# Draw nodes and labels
nx.draw_networkx_nodes(G_weighted, pos, node_color='lightblue', node_size=800)
nx.draw_networkx_labels(G_weighted, pos, font_size=12, font_weight='bold')

# Draw edge weights
edge_labels = nx.get_edge_attributes(G_weighted, 'weight')
nx.draw_networkx_edge_labels(G_weighted, pos, edge_labels)

plt.title(f"Shortest Path from {source} to {target} (highlighted in red)", 
          fontsize=14, fontweight='bold')
plt.axis('off')
plt.tight_layout()
plt.show()
```

## Minimum Spanning Tree

```python
# Find minimum spanning tree
mst = nx.minimum_spanning_tree(G_weighted)

print("Minimum Spanning Tree edges:")
total_weight = 0
for u, v, d in mst.edges(data=True):
    print(f"  {u}-{v}: weight = {d['weight']}")
    total_weight += d['weight']
    
print(f"\nTotal MST weight: {total_weight}")
```

```python
# Visualize MST
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))

# Original graph
pos = nx.spring_layout(G_weighted, seed=42)
nx.draw(G_weighted, pos, ax=ax1, with_labels=True, node_color='lightblue',
        node_size=800, font_size=12, font_weight='bold',
        edge_color='gray', width=2)
edge_labels = nx.get_edge_attributes(G_weighted, 'weight')
nx.draw_networkx_edge_labels(G_weighted, pos, edge_labels, ax=ax1)
ax1.set_title("Original Graph", fontsize=14, fontweight='bold')
ax1.axis('off')

# MST
nx.draw(mst, pos, ax=ax2, with_labels=True, node_color='lightgreen',
        node_size=800, font_size=12, font_weight='bold',
        edge_color='darkgreen', width=3)
mst_labels = nx.get_edge_attributes(mst, 'weight')
nx.draw_networkx_edge_labels(mst, pos, mst_labels, ax=ax2)
ax2.set_title("Minimum Spanning Tree", fontsize=14, fontweight='bold')
ax2.axis('off')

plt.tight_layout()
plt.show()
```

## Graph Coloring

```python
# Create a graph for coloring
G_color = nx.Graph()
G_color.add_edges_from([
    (1, 2), (1, 3), (1, 4),
    (2, 3), (2, 5),
    (3, 4), (3, 5), (3, 6),
    (4, 6),
    (5, 6)
])

# Greedy coloring
coloring = nx.greedy_color(G_color, strategy='largest_first')

print("Node coloring:")
for node, color in sorted(coloring.items()):
    print(f"  Node {node}: Color {color}")
    
num_colors = max(coloring.values()) + 1
print(f"\nNumber of colors used: {num_colors}")
```

```python
# Visualize coloring
plt.figure(figsize=(8, 6))
pos = nx.spring_layout(G_color, seed=42)

colors = ['red', 'blue', 'green', 'yellow', 'orange', 'purple']
node_colors = [colors[coloring[node]] for node in G_color.nodes()]

nx.draw(G_color, pos, with_labels=True, node_color=node_colors,
        node_size=800, font_size=14, font_weight='bold',
        edge_color='gray', width=2)

plt.title(f"Graph Coloring ({num_colors} colors)", fontsize=16, fontweight='bold')
plt.axis('off')
plt.tight_layout()
plt.show()
```

## PageRank Algorithm

PageRank measures the importance of nodes based on the link structure.

```python
# Create a directed graph representing web pages
Web = nx.DiGraph()

edges = [
    ('A', 'B'), ('A', 'C'), ('A', 'D'),
    ('B', 'C'), ('B', 'D'),
    ('C', 'A'),
    ('D', 'B'), ('D', 'C'),
    ('E', 'D'), ('E', 'B')
]

Web.add_edges_from(edges)

# Calculate PageRank
pagerank = nx.pagerank(Web, alpha=0.85)

print("PageRank scores:")
for node, score in sorted(pagerank.items(), key=lambda x: x[1], reverse=True):
    print(f"  {node}: {score:.4f}")
```

```python
# Visualize with PageRank
plt.figure(figsize=(10, 8))
pos = nx.spring_layout(Web, seed=42)

# Node size proportional to PageRank
node_sizes = [10000 * pagerank[node] for node in Web.nodes()]
node_colors = [pagerank[node] for node in Web.nodes()]

nx.draw(Web, pos, with_labels=True, node_size=node_sizes,
        node_color=node_colors, cmap='YlOrRd',
        font_size=12, font_weight='bold',
        edge_color='gray', width=2, arrows=True, arrowsize=20)

plt.title("Web Graph with PageRank (size = importance)", 
          fontsize=16, fontweight='bold')
plt.axis('off')
plt.tight_layout()
plt.show()
```

## Assortativity

Measures the tendency of nodes to connect to similar nodes.

```python
# Create networks with different assortativity
# Assortative: similar degree nodes connect
G_assort = nx.Graph()
G_assort.add_edges_from([
    (1, 2), (1, 3), (2, 3),  # High degree cluster
    (4, 5), (4, 6), (5, 6),  # High degree cluster
    (7, 8), (9, 10)           # Low degree pairs
])

# Disassortative: different degree nodes connect
G_disassort = nx.star_graph(5)  # Hub-and-spoke structure

# Calculate assortativity
assort_coef = nx.degree_assortativity_coefficient(G_assort)
disassort_coef = nx.degree_assortativity_coefficient(G_disassort)

print(f"Assortative network coefficient: {assort_coef:.3f}")
print(f"Disassortative network coefficient: {disassort_coef:.3f}")
```
